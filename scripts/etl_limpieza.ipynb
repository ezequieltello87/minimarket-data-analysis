```python
import pandas as pd
from pathlib import Path

# 1. Detectar automáticamente la raíz del proyecto ('practica minimarket')
# Si estamos parados dentro de 'scripts', sube un nivel ('..'). Si estamos en la raíz, se queda ahí.
raiz_proyecto = Path().resolve().parent if Path().resolve().name == 'scripts' else Path().resolve()

# 2. Construir las rutas relativas seguras usando la raíz
ruta_productos = raiz_proyecto / 'data' / 'raw' / 'productos_minimarket.csv'
ruta_ventas = raiz_proyecto / 'data' / 'raw' / 'ventas_minimarket.csv'
ruta_detalle = raiz_proyecto / 'data' / 'raw' / 'detalle_minimarket.csv'

# 3. Cargar cada uno en su propio DataFrame
df_productos = pd.read_csv(ruta_productos)
df_ventas = pd.read_csv(ruta_ventas)
df_detalle = pd.read_csv(ruta_detalle)

# 4. Verificar visualizando las primeras filas
display(df_productos.head())
display(df_ventas.head())
display(df_detalle.head())
       
```

```python

def auditar_espacios(df, nombre_df):
  print(f"--- Auditoría de espacios en: {nombre_df} ---")
  cols_texto = df.select_dtypes(include=['object', 'string']).columns

  for col in cols_texto:
    # Filtro con la regex exacta para extremos y dobles espacios internos
    sucios = df[df[col].astype(str).str.contains(r"^\s+|\s+$|\s{2,}", regex=True)]
    cantidad = len(sucios)

    print(f"Columna '{col}': {cantidad} registros con anomalías.")

    if cantidad > 0:
      display(sucios.head())
  print("\n")


# Ejemplo de uso para auditar tus DataFrames antes de limpiar:
auditar_espacios(df_productos, "df_productos")
auditar_espacios(df_ventas, "df_ventas")
```

```python
# 1. Integridad referencial
print("IDs producto huerfanos en detalle:", set(df_detalle['id_producto']) - set(df_productos['id_producto']))
print("IDs venta huerfanos en detalle:", set(df_detalle['id_venta']) - set(df_ventas['id_venta']))
print("IDs venta faltantes en detalle:", set(df_ventas['id_venta']) - set(df_detalle['id_venta']))

# 2. Consistencia de montos totales por venta
detalle_sum = df_detalle.groupby('id_venta')['subtotal'].sum().reset_index()
merged_ventas = pd.merge(df_ventas, detalle_sum, on='id_venta', suffixes=('_venta', '_detalle'))
mismatches_total = merged_ventas[merged_ventas['monto_total'] != merged_ventas['subtotal']]
print(f"Ventas donde monto_total != suma de subtotales: {len(mismatches_total)}")
```

```python
# Convertir la columna de texto a formato datetime real
df_ventas['fecha_hora'] = pd.to_datetime(df_ventas['fecha_hora'])
```

```python
df_ventas['id_cliente'] = df_ventas['id_cliente'].fillna('Consumidor Final')
```

```python
# 1. Normalizar textos clave en las tablas
df_productos['nombre'] = df_productos['nombre'].str.strip().str.title()
df_productos['categoria'] = df_productos['categoria'].str.strip().str.title()
df_ventas['medio_pago'] = df_ventas['medio_pago'].str.strip().str.title()
```

```python
df_productos.to_csv("../data/processed/productos_clean.csv", index=False)
df_ventas.to_csv("../data/processed/ventas_clean.csv", index=False)
df_detalle.to_csv("../data/processed/detalle_clean.csv", index=False)
print("¡Archivos limpios guardados exitosamente en data/processed/!")
```
